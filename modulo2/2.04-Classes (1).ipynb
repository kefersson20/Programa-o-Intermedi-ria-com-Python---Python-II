#### Q1.
Vamos criar aos pouquinhos uma classe Evento que registra eventos de um calendário. Pra começar, nessa questão você deve:

* Criar uma classe chamada Evento
* Adicionar um atributo de classe `total_eventos` que será usado para contar o número total de eventos (instâncias de classe) criadas.
* Implemente o método construtor que deve receber os parâmetros `titulo` (`string`), `data_hora` (`datetime`), `descrição` (`string`). Crie e inicialize um atributo de instância para cada um dos parâmetros citados.
* O construtor também deve inicializar um atributo de instância `is_concluido = False`. Esse atributo será utilizado mais à frente. 
* No construtor você também deve incrementar o atributo de classe `total_eventos` em 1 a cada nova instância criada.

Teste o seu código criando duas diferentes instâncias de evento (preencha os atributos como quiser) e imprimindo todos os seus atributos (de classe e de instância).

> Consulte [este tutorial biblioteca `datetime`](https://www.w3schools.com/python/python_datetime.asp) caso não esteja familiarizado.

```python

from datetime import datetime, timedelta


class Evento:
    total_eventos = 0

    def __init__(self, titulo: str, data_hora: datetime, descricao: str):
        self.titulo = titulo
        self.data_hora = data_hora
        self.descricao = descricao
        self.is_concluido = False

        Evento.total_eventos += 1


# Teste com duas instâncias
evento1 = Evento(
    "Reunião de equipe",
    datetime(2026, 10, 10, 14, 0),
    "Reunião para discutir o projeto."
)

evento2 = Evento(
    "Entrega do trabalho",
    datetime(2026, 10, 5, 18, 0),
    "Entrega da atividade de Python."
)

# Imprimindo os atributos de instância
print(evento1.titulo)
print(evento1.data_hora)
print(evento1.descricao)
print(evento1.is_concluido)

print(evento2.titulo)
print(evento2.data_hora)
print(evento2.descricao)
print(evento2.is_concluido)

# Imprimindo o atributo de classe
print("Total de eventos:", Evento.total_eventos)
```

* Um método de classe `num_eventos()` que retorna o valor do atributo de classe `total_eventos`.

Para o teste, use a função nativa do Python [`isinstance`](https://www.w3schools.com/python/ref_func_isinstance.asp).

Para testar sua classe atualizada:
* Crie uma instância de Evento com valor passado e chame o método `isConcluido()` para a instância criada. Em seguida imprima o atributo `is_concluido`.

```python

from datetime import datetime


class Evento:
    total_eventos = 0

    def __init__(self, titulo: str, data_hora: datetime, descricao: str):
        self.titulo = titulo
        self.data_hora = data_hora
        self.descricao = descricao
        self.is_concluido = False

        Evento.total_eventos += 1

    def isConcluido(self):
        if self.data_hora < datetime.now():
            self.is_concluido = True

    @classmethod
    def num_eventos(cls):
        return cls.total_eventos

    @staticmethod
    def valida_evento(nome, data_hora, descricao):
        return (
            isinstance(nome, str)
            and isinstance(data_hora, datetime)
            and isinstance(descricao, str)
        )


# Teste do evento passado
evento3 = Evento(
    "Prova de Python",
    datetime.now() - timedelta(days=1),
    "Avaliação de programação."
)

evento3.isConcluido()
print("Evento concluído:", evento3.is_concluido)

# Teste do método de classe
print("Total de eventos:", Evento.num_eventos())

# Testes do método estático
print(Evento.valida_evento(
    "Palestra", datetime.now(), "Palestra de tecnologia."
))

print(Evento.valida_evento(
    123, "data inválida", False
))
```

```python

from datetime import datetime, timedelta


class Evento:
    total_eventos = 0

    def __init__(self, titulo: str, data_hora: datetime, descricao: str):
        self.titulo = titulo
        self.data_hora = data_hora
        self.descricao = descricao
        self.is_concluido = False

        Evento.total_eventos += 1

    def isConcluido(self):
        if self.data_hora < datetime.now():
            self.is_concluido = True

    @classmethod
    def num_eventos(cls):
        return cls.total_eventos

    @staticmethod
    def valida_evento(nome, data_hora, descricao):
        return (
            isinstance(nome, str)
            and isinstance(data_hora, datetime)
            and isinstance(descricao, str)
        )

    def __str__(self):
        return (
            f"Evento: {self.titulo}, "
            f"Data: {self.data_hora}, "
            f"Descrição: {self.descricao}, "
            f"Concluido: {self.is_concluido}"
        )

    def __eq__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora == outro.data_hora

    def __ne__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora != outro.data_hora

    def __lt__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora < outro.data_hora

    def __le__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora <= outro.data_hora

    def __gt__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora > outro.data_hora

    def __ge__(self, outro):
        if not isinstance(outro, Evento):
            return NotImplemented
        return self.data_hora >= outro.data_hora


# Criando dois eventos com datas diferentes
evento1 = Evento(
    "Reunião",
    datetime(2026, 10, 10, 14, 0),
    "Reunião da equipe."
)

evento2 = Evento(
    "Entrega do projeto",
    datetime(2026, 10, 15, 18, 0),
    "Entrega final do projeto."
)

# Imprimindo as instâncias
print(evento1)
print(evento2)

# Comparando os eventos
print("==", evento1 == evento2)
print("!=", evento1 != evento2)
print("<", evento1 < evento2)
print("<=", evento1 <= evento2)
print(">", evento1 > evento2)
print(">=", evento1 >= evento2)
```
